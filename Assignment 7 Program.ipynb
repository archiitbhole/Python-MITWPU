```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Take CSV file name from user
file_name = input("Enter CSV file name: ")

# Load the CSV dataset
df = pd.read_csv(file_name)

# Display the dataset
print("\nDataset:")
print(df)

# 1. Line Plot - Python marks
plt.figure(figsize=(7, 4.5))
plt.plot(df["Student"], df["Python"], marker="o", label="Python Marks")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.legend()
plt.grid(True)
plt.show()

# 2. Bar Plot - Maths marks
plt.figure(figsize=(7, 4.5))
plt.bar(df["Student"], df["Maths"], label="Maths Marks")
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.legend()
plt.show()

# 3. Scatter Plot - Maths vs Python
plt.figure(figsize=(7, 4.5))
plt.scatter(df["Maths"], df["Python"], s=70, label="Students")
plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.legend()
plt.grid(True)
plt.show()

# 4. Correlation Matrix and Heatmap
correlation = df[["Maths", "Python", "Attendance"]].corr()
print("\nCorrelation Matrix:")
print(correlation)

plt.figure(figsize=(6.5, 5))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()

# 5. Histogram - Python marks
plt.figure(figsize=(7, 4.5))
plt.hist(df["Python"], bins=5, edgecolor="black", label="Python Marks")
plt.title("Histogram of Python Marks")
plt.xlabel("Python Marks")
plt.ylabel("Number of Students")
plt.legend()
plt.show()

# 6. Box Plot - Maths and Python marks
plt.figure(figsize=(7, 4.5))
plt.boxplot([df["Maths"], df["Python"]], tick_labels=["Maths", "Python"])
plt.title("Box Plot of Maths and Python Marks")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()

# 7. Pie Chart - Students per City
city_count = df["City"].value_counts()
print("\nNumber of Students per City:")
print(city_count)

plt.figure(figsize=(6, 6))
plt.pie(city_count, labels=city_count.index, autopct="%1.1f%%", startangle=90)
plt.title("Students from Each City")
plt.legend(title="City", loc="upper right")
plt.show()

# 8. Averages
print("\nAverage Marks and Attendance:")
print("Average Maths Marks :", round(df["Maths"].mean(), 2))
print("Average Python Marks:", round(df["Python"].mean(), 2))
print("Average Attendance  :", round(df["Attendance"].mean(), 2))


#INPUT:
'''
Student,Maths,Python,Attendance,City
Aarav,78,85,90,Pune
Diya,88,92,95,Mumbai
Rohan,65,70,80,Pune
Sneha,92,89,97,Nagpur
Kabir,55,60,72,Mumbai
Ananya,81,78,88,Pune
Vihaan,70,75,85,Nashik
Isha,95,98,99,Mumbai

Enter CSV file name: sample_data_assignment_07.csv
'''

#OUTPUT:
'''
Dataset:
  Student  Maths  Python  Attendance    City
0   Aarav     78      85          90    Pune
1    Diya     88      92          95  Mumbai
2   Rohan     65      70          80    Pune
3   Sneha     92      89          97  Nagpur
4   Kabir     55      60          72  Mumbai
5  Ananya     81      78          88    Pune
6  Vihaan     70      75          85  Nashik
7    Isha     95      98          99  Mumbai

Correlation Matrix:
               Maths    Python  Attendance
Maths       1.000000  0.964271    0.988186
Python      0.964271  1.000000    0.982792
Attendance  0.988186  0.982792    1.000000

Number of Students per City:
City
Pune      3
Mumbai    3
Nagpur    1
Nashik    1
Name: count, dtype: int64

Average Marks and Attendance:
Average Maths Marks : 78.0
Average Python Marks: 80.88
Average Attendance  : 88.25
'''
```
